### Data Sync Utility
One-time and reusable transfers from Unity Catalog Volumes into this Git folder for version control.

```python
## Define the fucntion to move the files.
def file_mover(table_name, dst):
    df=spark.table(table_name)
    df.toPandas().to_csv(dst, index=False)
    print(f"Exported: {os.path.basename(dst)}") # final message when completed
```
