### In the following lines, we will write scripts to move files from one location to another

```python
import shutil
import os


def copy_to_dest_folder(src, dest):
    if not os.path.exists(src):
        print(f"Source not found, skipped: {src}")
        return
    shutil.copy(src, dest)
    print(f"Copied: {os.path.basename(dest)}")


def move_anything(src, dst):
    if not os.path.exists(src):
        print(f"Source not found, skipped: {src}")
        return
    if os.path.isdir(src):
        shutil.copytree(src, dst)
        print(f"Folder copied: {os.path.basename(dst)}")
    else:
        shutil.copy(src, dst)
        print(f"File copied: {os.path.basename(dst)}")

```
