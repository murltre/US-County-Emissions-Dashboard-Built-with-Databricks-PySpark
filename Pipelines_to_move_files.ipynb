### Data Sync Utility
One-time and reusable transfers from Unity Catalog Volumes into this Git folder for version control.

```python
## Define the fucntion to move the files.
def file_mover(table_name, dst):
    df=spark.table(table_name)
    df.toPandas().to_csv(dst, index=False)
    print(f"Exported: {os.path.basename(dst)}") # final message when completed
```

```python
# Define the paths: we have a list of tuples -- each tuples has 2 elements - source file and destination of the file.
dst_folder = "/Workspace/Users/mourlayetraore@mtanalytics.onmicrosoft.com/Databricks & PySpark/Datasets/"
tables_to_move =[
    ("us_emissions_catalog_8_5.raw.emissions_data", dst_folder + "us_emission_rawdata.csv"),
    ("us_emissions_catalog_8_5.raw.emission_df2", dst_folder + "emission_cleandata"),
]
```

```python
## Here we run a for look to execute the function and move th files.
for x, y in tables_to_move:
    file_mover(x, y)
```
