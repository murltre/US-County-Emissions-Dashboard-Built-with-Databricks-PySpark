### Data Sync Utility
One-time and reusable transfers from Unity Catalog Volumes into this Git folder for version control.

```python
import os
def export_table_to_git_folder(table_name, dst):
    df=spark.table(table_name)
    df.toPandas().to_csv(dst, index=False)
    print(f"Exported: {os.path.basename(dst)}")

dest_folder = "/Workspace/Users/mourlayetraore@mtanalytics.onmicrosoft.com/Databricks & PySpark/Datasets/"
```

```python
tables_to_copy = [
    ("us_emissions_catalog_8_5.raw.emissions_data", dest_folder + "emissions_rawdata.csv"),
    ("us_emissions_catalog_8_5.raw.emission_df2", dest_folder + "emission_df2.csv"),
]

for table_name, dst in tables_to_copy:
    export_table_to_git_folder(table_name, dst)
```
